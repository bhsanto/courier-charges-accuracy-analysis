```python
import pandas as pd
```

```python
#Loading datasets
order_report = pd.read_csv('Order Report.csv')
sku_master = pd.read_csv('SKU Master.csv')
pincode_mapping = pd.read_csv('pincodes.csv')
courier_invoices = pd.read_csv('Invoice.csv')
courier_company_rates = pd.read_csv('Courier Company - Rates.csv')
```

```python
#Visualizing the datasets first five rows (default)
print("ORDER REPORT: ")
print(order_report.head())

print("\nSKU MASTER: ")
print(sku_master.head())

print("\nPINCODE MAPPING: ")
print(pincode_mapping.head())

print("\nCOURIER INVOICES: ")
print(courier_invoices.head())

print("\nCOURIER COMPANY RATES: ")
print(courier_company_rates.head())
```

```python
#Checking Missing Values in the Dataframes
print("\nMISSING VALUES IN ORDER REPORT: ")
print(order_report.isnull().sum())

print("\nMISSING VALUES IN SKU MASTER: ")
print(sku_master.isnull().sum())

print("\nMISSING VALUES IN PINCODE MAPPING: ")
print(pincode_mapping.isnull().sum())

print("\nMISSING VALUES IN COURIER INVOICES: ")
print(courier_invoices.isnull().sum())

print("\nMISSING VALUES IN COURIER COMPANY RATES: ")
print(courier_company_rates.isnull().sum())
```

```python
#Let's clean the data!
#Removing unnamed columns from the dataframes with missing values
order_report = order_report.drop(columns=['Unnamed: 3','Unnamed: 4'])
sku_master = sku_master.drop(columns=['Unnamed: 2', 'Unnamed: 3','Unnamed: 4'])
pincode_mapping = pincode_mapping.drop(columns=['Unnamed: 3', 'Unnamed: 4'])
```

```python
#Visualize dataframes after dropping the columns with missing values
print("\nORDER REPORT AFTER CLEANING: ")
print(order_report.head())

print("\nSKU MASTER AFTER CLEANING: ")
print(sku_master.head())

print("\nPINCODE MAPPING AFTER CLEANING: ")
print(pincode_mapping.head())
```

```python
#Merging ORDER REPORT and SKU MASTER dataframes based on the common SKU column
merged_df = pd.merge(order_report, sku_master, on='SKU')
print(merged_df.head())
```

```python
#Renaming the ExternOrderNo to Order Id in the mergerd_df
merged_df = merged_df.rename(columns={'ExternOrderNo': 'Order Id'})
```

```python
#Removing duplicate rows from Customer Pincode column in PINCODE MAPPING dataframe
abc_courier = pincode_mapping.drop_duplicates(subset=['Customer Pincode'])

#Merging COURIER INVOICES and PINCODE MAPPING dataframes
courier_abc = courier_invoices[['Order ID', 'Customer Pincode', 'Type of Shipment']]
pincode = courier_abc.merge(abc_courier, on='Customer Pincode')
print(pincode.head())
```
