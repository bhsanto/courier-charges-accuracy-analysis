```python
import pandas as pd
```

```python
#Loading datasets
order_report = pd.read_csv('Order Report.csv')
sku_master = pd.read_csv('SKU Master.csv')
pincode_mapping = pd.read_csv('pincodes.csv')
courier_invoices = pd.read_csv('Invoice.csv')
courier_company_rates = pd.read_csv('Courier Company - Rates.csv')
```

```python
#Visualizing the datasets first five rows (default)
print("ORDER REPORT: ")
print(order_report.head())

print("\nSKU MASTER: ")
print(sku_master.head())

print("\nPINCODE MAPPING: ")
print(pincode_mapping.head())

print("\nCOURIER INVOICES: ")
print(courier_invoices.head())

print("\nCOURIER COMPANY RATES: ")
print(courier_company_rates.head())
```

```python
#Checking Missing Values in the Dataframes
print("\nMISSING VALUES IN ORDER REPORT: ")
print(order_report.isnull().sum())

print("\nMISSING VALUES IN SKU MASTER: ")
print(sku_master.isnull().sum())

print("\nMISSING VALUES IN PINCODE MAPPING: ")
print(pincode_mapping.isnull().sum())

print("\nMISSING VALUES IN COURIER INVOICES: ")
print(courier_invoices.isnull().sum())

print("\nMISSING VALUES IN COURIER COMPANY RATES: ")
print(courier_company_rates.isnull().sum())
```

```python
#Let's clean the data!
#Removing unnamed columns from the dataframes with missing values
order_report = order_report.drop(columns=['Unnamed: 3','Unnamed: 4'])
sku_master = sku_master.drop(columns=['Unnamed: 2', 'Unnamed: 3','Unnamed: 4'])
pincode_mapping = pincode_mapping.drop(columns=['Unnamed: 3', 'Unnamed: 4'])
```

```python
#Visualize dataframes after dropping the columns with missing values
print("\nORDER REPORT AFTER CLEANING: ")
print(order_report.head())

print("\nSKU MASTER AFTER CLEANING: ")
print(sku_master.head())

print("\nPINCODE MAPPING AFTER CLEANING: ")
print(pincode_mapping.head())
```

```python
#Merging ORDER REPORT and SKU MASTER dataframes based on the common SKU column
merged_df = pd.merge(order_report, sku_master, on='SKU')
print(merged_df.head())
```

```python
#Renaming the ExternOrderNo to Order Id in the mergerd_df
merged_df = merged_df.rename(columns={'ExternOrderNo': 'Order ID'})
```

```python
print(merged_df.head())
```

```python
#Removing duplicate rows from Customer Pincode column in PINCODE MAPPING dataframe
abc_courier = pincode_mapping.drop_duplicates(subset=['Customer Pincode'])

#Merging COURIER INVOICES and PINCODE MAPPING dataframes
courier_abc = courier_invoices[['Order ID', 'Customer Pincode', 'Type of Shipment']]
pincode = courier_abc.merge(abc_courier, on='Customer Pincode')
print(pincode.head())
```

```python
#Merging the PINCODE with main dataframe
merged2_df = merged_df.merge(pincode, on='Order ID')
print(merged2_df.head())
```

```python
#Now calculating the weight 'Weight (g)' in kilograms dividing by 1000
merged2_df['Weights (Kg)'] = merged2_df['Weight (g)'] / 1000
print(merged2_df.head())
```

```python
#Calculating the weight slabs
def weight_slab(weight):
  i = round(weight % 1, 1)
  if i == 0.0:
    return weight
  elif i > 0.5:
    return int(weight) + 1.0
  else:
    return int(weight) + 0.5

merged2_df['Weight Slab (KG)'] = merged2_df['Weights (Kg)'].apply(weight_slab)
courier_invoices['Weight Slab Charged by Courier Company'] = (courier_invoices['Charged Weight']).apply(weight_slab)
```

```python
#Renaming the columns
courier_invoices = courier_invoices.rename(columns={'Zone':'Delivery Zone Charged by Courier Company'})
merged2_df = merged2_df.rename(columns={'Zone':'Delivery Zone As Per ABC'})
merged2_df = merged2_df.rename(columns={'Weight Slab (KG)': 'Weight Slab As Per ABC'})
```

```python
#Calculating the expected charges

total_expected_charge = []

for _, row in merged2_df.iterrows():
  fwd_category = 'fwd_' + row['Delivery Zone As Per ABC']
  fwd_fixed = courier_company_rates.at[0, fwd_category + '_fixed']
  fwd_additional = courier_company_rates.at[0, fwd_category + '_additional']
  rto_category = 'rto_' + row['Delivery Zone As Per ABC']
  rto_fixed = courier_company_rates.at[0, rto_category + '_fixed']
  rto_additional = courier_company_rates.at[0, rto_category + '_additional']

  weight_slab = row['Weight Slab As Per ABC']

  if row['Type of Shipment'] == 'Forward charges':
    additional_weight = max(0, (weight_slab - 0.5) / 0.5)
    total_expected_charge.append(fwd_fixed + additional_weight * fwd_additional)
  elif row['Type of Shipment'] == 'Forward and RTO charges':
    additional_weight = max(0, (weight_slab - 0.5) / 0.5)
    total_expected_charge.append(fwd_fixed + additional_weight * (fwd_additional + rto_additional))
  else:
    total_expected_charge.append(0)

merged2_df['Expected Charge as per ABC'] = total_expected_charge
print(merged2_df.head())
```

```python
#Merging it with COURIER INVOICES toget the final dataframe
merged_output = merged2_df.merge(courier_invoices, on='Order ID')
print(merged_output.head())
```

```python
#Calculating the differences between charges and expected charges for each order
df_diff = merged_output
df_diff['Difference (Rs.)'] = df_diff['Billing Amount (Rs.)'] - df_diff['Expected Charge as per ABC']

df_new = df_diff[['Order ID', 'Billing Amount (Rs.)', 'Expected Charge as per ABC', 'Difference (Rs.)']]
print(df_new.head())
```
